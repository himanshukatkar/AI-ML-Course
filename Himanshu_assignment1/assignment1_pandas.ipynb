**Q12. Replace the text values NA, N/A, null and - with proper missing values.**

```python
missing_tokens = ["NA", "na", "N/A", "n/a", "NULL", "Null", "null", "-", "--"]
df = df.replace(missing_tokens, np.nan)
df.isna().sum().sum()
```

*The total missing-value count went up because placeholder text like `NA`, `N/A`, `null`, and `-` is now treated as a real missing value instead of a string. (Other junk tokens like `?` and `missing` are cleared out later when each numeric column is converted, since `pd.to_numeric(errors="coerce")` turns any leftover non-numeric text into NaN too.)*

**Q25. Convert the hours logged column to a number.**

```python
df['hours_logged'] = df['hours_logged'].astype(str).str.replace('hrs', '', regex=False).str.strip()
df['hours_logged'] = pd.to_numeric(df['hours_logged'], errors='coerce')
df['hours_logged'].dtype
```

*`hours_logged` is now a true numeric column; any leftover junk text (like `?` or `missing`) became NaN instead of crashing the conversion.*

**Q26. Convert the billing rate column to a number. Remove the $ sign first.**

```python
df['billing_rate_(usd)'] = (
    df['billing_rate_(usd)'].astype(str)
    .str.replace('$', '', regex=False)
    .str.replace('USD', '', regex=False)
    .str.strip()
)
df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'], errors='coerce')
df['billing_rate_(usd)'].dtype
```

*Billing rate is now numeric after stripping the `$` sign and stray `USD` prefixes some rows had.*

**Q27. Convert the CSAT score column to a number.**

```python
df['csat_score'] = pd.to_numeric(df['csat_score'], errors='coerce')
df['csat_score'].dtype
```

*CSAT score is now numeric, with invalid entries (like the stray literal `CSAT Score` text from a duplicated header row) turned into NaN.*

**Q28. Print the average, the smallest and the largest value of hours logged.**

```python
print("Average:", df['hours_logged'].mean())
print("Smallest:", df['hours_logged'].min())
print("Largest:", df['hours_logged'].max())
```

*Most tickets take a modest number of hours, but the maximum shows at least one ticket logged a very large number of hours.*

**Q30. Show the 5 tickets with the highest billable amount.**

```python
df['billable_amount_usd'] = (
    df['billable_amount_usd'].astype(str)
    .str.replace('$', '', regex=False)
    .str.replace('USD', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)
df['billable_amount_usd'] = pd.to_numeric(df['billable_amount_usd'], errors='coerce')
df.sort_values('billable_amount_usd', ascending=False).head(5)
```

*Billable amount had `$`, `USD`, and thousands-separator commas mixed in, so it's cleaned to numeric here before finding the top 5 highest-billed tickets.*
